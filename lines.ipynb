```python
my_line_file = 'input_svgs/too_many_test.txt'
output_file = 'svgs_lines/too_many_test_lines.txt'

with open(my_line_file, 'r') as file:
    my_long_line = file.read()
print(my_long_line) 
```

```python
print(len(my_long_line))
my_long_line
```

```python
start = '<'
end = '>'
```

```python
def write_line(out_file, line):
    file = open(out_file, "a")
    file.write(line + "\n")
    file.close()
```

```python
len_line = len(my_long_line)


while(len_line > 1):
    index_start = my_long_line.find(start)
    index_end = my_long_line.find(end)
    if index_start != -1 and index_end != -1:
        short_line = my_long_line[index_start + len(start):index_end+1]
        print(short_line)
        write_line(output_file, short_line)
        my_long_line = my_long_line.replace(short_line, "")
        len_line = len(my_long_line)
        print(len_line)
    else:
        print('Nope!')
```
