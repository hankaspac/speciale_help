```python
import numpy as np
import altair as alt
import pandas as pd 
import os
import cv2
```

```python
input_file = 'svgs_lines/too_many_test_lines.txt'
output_file = 'extracted_lines/wrong_chart_maps_colors_extract.txt'
```

```python
def write_line(out_file, line):
    file = open(out_file, "a")
    file.write(line + "\n")
    file.close()
```

```python
with open(input_file, 'r') as file:
    for line in file:
        if "background-color:" in line:
            write_line(output_file, line)
        if "<path aria-label" in line:
            write_line(output_file, line)
        if "role-legend" in line:
            write_line(output_file, line)    
        if "path transform" and "fill=" in line:  
            write_line(output_file, line)    
```
