```python
import numpy as np
import altair as alt
import pandas as pd 
import os
import cv2

```

```python
alt.renderers.enable("browser")
#alt.renderers.enable("html")
```

```python
games = pd.read_excel("data/board_games.xlsx", sheet_name='board_games')
games 
```

```python
alt.Chart(games).mark_point(filled=True).encode(
    x='Title',
    y='Nr_boxes', #'Type',
    color='Max_players:N',
    size='Campagne',  #'Rating_avg:Q',
    shape='Coop',
    tooltip=['Title', 'Type', 'Max_players', 'Nr_boxes', 'Year', 'Rating_avg']
).interactive()
```

```python
alt.Chart(games).mark_point().encode(
    x='Title',
    size='Max_players',
    color='Type',
    y='Nr_boxes',
    shape='Coop'
).interactive()
```

```python
alt.Chart(games).mark_point(filled=True).encode(
    x='Campagne',
    y='Type',
    color='Type',
    shape='Coop',
    tooltip=['Title', 'Type', 'Max_players', 'Nr_boxes', 'Year', 'Rating_avg']
).interactive()
```
